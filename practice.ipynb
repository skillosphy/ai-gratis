```python
import numpy as np

# Linear regression
# prediction
def predict(w, x, b):
    return w*x+b

# MSE
def compute_cost (x, y, w, b):
    n = len(x) 
    y_hat = predict(w, x, b)
    errors = y_hat - y
    return (1/(2*n) * np.sum(errors**2))


def compute_gradient (x, y, w, b)
    n = len(y)
    errors = predict(w, x, b) - y
    dj_dw = (1/n) * np.sum(errors * x)
    dj_db = (1/n) * np.sum(errors)

    return dj_dw, dj_db

def grad_desc(x, y, w_init, b_init, lr, iterations)
    w,b = w_init, b_init
    cost_history = []

    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(x, y, w, b)
        w = w - lr(dj_dw)
        b = b - lr(dj_db)
        cost_history.append(compute_cost(x,y,w,b))

    return w,b,cost_history

# Logistic Regression
def sigmoid(z):
    return 1/(1 + np.exp(-z))

def predict_proba(x, w, b):
    return sigmoid(w*x+b)

def predict_class(w,x,b, threshold = 0.5):
    return (predict_proba(x, w, b) >= threshold).astype(int)


def compute_log_loss(x, y, w, b):
    n = len(y)
    y_hat = predict_proba(x, w,b)
    return -(1/n) *np.sum(y*np.log(y_hat) + (1-y)*(np.log(1-y_hat)))


# Regularization will use @ instead of * cuz x is matrix
def compute_cost_l2(x, y, w, b, lmb):
    m = len(y)
    error = (x @ w +b) - y
    mse = 1/(2*m) * np.sum(error**2)
    penalty = (lmb/2*m * np.sum(w**2))
    return mse + penalty

def compute_gratis_l2 (x, y, w, b, lmb):
    m = len(y)
    error = (x @ w + b) - y
    dj = 1/m * (x.T @ error) + (lmb/m)*w
    db = 1/m * (x.T @ error)

    return dj, db














```
