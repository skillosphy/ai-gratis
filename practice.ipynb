```python
import numpy as np

# Linear regression
# prediction
def predict(w, x, b):
    return w*x+b

# MSE
def compute_cost (x, y, w, b):
    n = len(x) 
    y_hat = predict(w, x, b)
    errors = y_hat - y
    return (1/(2*n) * np.sum(errors**2))


def compute_gradient (x, y, w, b)
    n = len(y)
    errors = predict(w, x, b) - y
    dj_dw = (1/n) * np.sum(errors * x)
    dj_db = (1/n) * np.sum(errors)

    return dj_dw, dj_db

def grad_desc(x, y, w_init, b_init, lr, iterations)
    w,b = w_init, b_init
    cost_history = []

    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(x, y, w, b)
        w = w - lr(dj_dw)
        b = b - lr(dj_db)
        cost_history.append(compute_cost(x,y,w,b))

    return w,b,cost_history

# Logistic Regression
def sigmoid(z):
    return 1/(1 + np.exp(-z))

def predict_proba(x, w, b):
    return sigmoid(w*x+b)

def predict_class(w,x,b, threshold = 0.5):
    return (predict_proba(x, w, b) >= threshold).astype(int)


def compute_log_loss(x, y, w, b):
    n = len(y)
    y_hat = predict_proba(x, w,b)
    return -(1/n) *np.sum(y*np.log(y_hat) + (1-y)*(np.log(1-y_hat)))


# Regularization will use @ instead of * cuz x is matrix
def compute_cost_l2(x, y, w, b, lmb):
    m = len(y)
    error = (x @ w +b) - y
    mse = 1/(2*m) * np.sum(error**2)
    penalty = (lmb/2*m * np.sum(w**2))
    return mse + penalty

def compute_gratis_l2 (x, y, w, b, lmb):
    m = len(y)
    error = (x @ w + b) - y
    dj = 1/m * (x.T @ error) + (lmb/m)*w
    db = 1/m * (x.T @ error)

    return dj, db



# Stuff missing
def compute_gradients_logisitic(x, y, w, b)
    n = len(y)
    y_hat = predict_proba(w,x,b)

    dj_dw = (1/n) * np.sum(y_hat - y) * x
    dj_db = (1/n) * np.sum(y_hat - y)

    return dj_dw, dj_db

def logistic_grad_desc(x, y, w_init, b_init, lr, iterations):
    w,b = w_init, b_init
    cost_history = []

    for i in range(iterations):
        dj_dw, dj_db = compute_gradients_logisitic(x, y, w, b)
        w = w - lr * dj_dw
        b = b - lr * dj_db
        cost_history.append(compute_log_loss(x,y,w,b))

    return w,b,cost_history

```

```python
# Evaluation Metrics

import numpy as np
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay, accuracy_score, precision_score, recall_score, f1_score, classification_report)
import matplotlib.pyplot as plt

y_true = np.array([1, 0, 1, 1, 0, 0, 1, 0, 0, 1])
y_pred = np.array([1, 0, 0, 1, 0, 1, 1, 0, 0, 0])

#Manual
# 1. Compute Confusion Matrix Components Manually
TP = np.sum((y_true == 1) & (y_pred == 1))  # True Positives: Actual 1, Predicted 1
TN = np.sum((y_true == 0) & (y_pred == 0))  # True Negatives: Actual 0, Predicted 0
FP = np.sum((y_true == 0) & (y_pred == 1))  # False Positives: Actual 0, Predicted 1 (Type I error)
FN = np.sum((y_true == 1) & (y_pred == 0))  # False Negatives: Actual 1, Predicted 0 (Type II error)

print(f"TP: {TP}, TN: {TN}, FP: {FP}, FN: {FN}")

# 2. Compute Metrics Manually
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0  # Out of all predicted positive, how many were right?
recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0     # Out of all actual positive, how many did we catch?
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

print(f"Accuracy: {accuracy:.2f}, Precision: {precision:.2f}, Recall: {recall:.2f}, F1: {f1_score:.2f}")


#scikit implemenation

print("Accuracy", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
# print("F1 Score", f1_score(y_true, y_pred)) dosent work for some frigging reason



print("Classification Report", classification_report(y_true, y_pred, target_names=["Not Spam", "Spam"]))
```
